**We will first solve this scnerio:**
- A user can create and view 2D coordinates
- A user can find out the distance between 2 coordinates
- A user can find find the distance of a coordinate from origin
- A user can check if a point lies on a given line
- A user can find the distance between a given 2D point and a given line

```python
class Point:

  def __init__(self, x, y):
    self.x_cor = x
    self.y_cor = y

  def __str__(self):
    return '<{},{}>'.format(self.x_cor, self.y_cor) # we want to display points like <0,0> . Thats why we write this.

  def eculidean_distance(self, other): # here when object created it will be recieved as self and we are using the "other" to store 2nd point
    return ((self.x_cor - other.x_cor)**2 + (self.y_cor - other.y_cor)**2)**0.5 # This is the formula of euclidean distance


  def distance_from_origin(self): # here we use only one argument which is self which we recieve as object which contain our point and we need to calculate its distance from the origin which is (0,0)
    #return ((self.x_cor - 0)**2 + (self.y_cor - 0)**2)**0.5
    return self.eculidean_distance(Point(0,0)) # This is a smart way. You are calling your method and creating an object of the class inside the class.


class Line:
  def __init__(self, X, Y, Z):
    self.A = X
    self.B = Y
    self.C = Z

  def __str__(self):
    return '{}x + {}y + {}'.format(self.A, self.B, self.C)

  def point_on_line(line, point): # here we are using the word line and point instead of self and other
    if line.A*point.x_cor + line.B*point.y_cor + line.C == 0: # this is Maths if we put the cordinates in the equation of line (Ax+By+C=0) and if answer is 0 then it means the point lies on the line.
      return "Point lies on the line"
    else:
      return "Point does not lies on the line"


  def shortest_distance(line, point): # this is the 5th question
    return abs(line.A*point.x_cor + line.B*point.y_cor + line.C)/(line.A**2 + line.B**2)**0.5 # This the formula of shortest distance between a line and a point.
```

```python
class person:
  def __init__(self):
    self.name = 'Muneeb'
    self.gender = 'Male'
```

```python
person() # Here the object is created
p = person() # Here I again created the object and store its REFERENCE or ADDRESS in the p variable.
             # So p itself is not the object infact it is the reference variable that points out to the address of the object

q = p    # Here q will also points to same object So I created two reference variables for the same object

print(id(p))
print(id(q))
```

### **Pass by Reference**

- We can pass the object as argument to the function and also we can return the object from the Function.

- Infact you are not passing the object but the reference of the object.
In below code you can clearly see you are pass the the reference variable of the object as the argument of the function as well as you are returning the reference of the object.

- The object is stored in the fixed memory address and all the variables are pointing out to that same memory address and agian changing from one side will reflect the same change everywhere no matter from which reference you are accessing the object.

```python
class Person:
  def __init__(self, name , gender):
    self.name = name
    self.gender = gender

# Now I will create a Function because it is outside the class

def greet(person):
  print(f"My name is {person.name} and my gender is {person.gender}.")
  p1 = Person("Ali", "Male")
  print(id(p1))
  return p1


p = Person("Muneeb" , "Male")

greet(p) # Here the argument is the Function.

x = greet(p)
print(x.name)
print(x.gender)
print(id(x))
```

```python
# Static Variables

class Atm:

  __counter = 1 # We have to make it private

  def __init__(self):
    print(id(self))
    self.pin = ''
    self.__balance = 0
    self.customer_id = Atm.__counter
    Atm.__counter = Atm.__counter + 1

    #self.menu()
  @staticmethod   #we write @staticmethod DECORATOR to identify the class methods. SO i dont need to create object to access them. and also called as Utility Functions.
  def get_counter(): # Look we didnt write self in the bracket because it doesnt need of it but it we try
    return Atm.__counter # to access it through object it will throw error so we access it through class name in the below cell.


  def get_balance(self):       # THIS IS GETTER FUNCTION
    return self.__balance

  def set_balance(self,new_value):  # THIS IS SETTER FUNCTION
    if type(new_value) == int:
      self.__balance = new_value
    else:
      print('beta bahot maarenge')

```

### **Points to remember about static**

- Static attributes are created at class level.
- Static attributes are accessed using ClassName.
- Static attributes are object independent. We can access them without creating instance (object) of the class in which they are defined.
- The value stored in static attribute is shared between all instances(objects) of the class in which the static attribute is defined.

```python
class Lion:
  __water_source="well in the circus"

  def __init__(self,name, gender):
      self.__name=name
      self.__gender=gender

  def drinks_water(self):
      print(self.__name,
      "drinks water from the",Lion.__water_source)

  @staticmethod
  def get_water_source():
      return Lion.__water_source

simba=Lion("Simba","Male")
simba.drinks_water()
print( "Water source of lions:",Lion.get_water_source())
```
