## **Inheritance:**
(Is a) relationship between two classes.
Inheritance means the a class (Child Class) can access the attributes and the methods of the other class( Parent Class). Means the object of child class can access the methods and attributes of the parent class from which it is inherited.

**Suppose you are making an app like Udemy where you know we have Students and Instructor.** <br>
 The login and register are the funtionalities that boht students and instructor have so instead to writring the same code of login in both classes, <br> We will create a Parent class (User Class) for login method that contains the login code and then we will create two child classes one for instructor that contain its methods like create a course or reply to questions and one class for students that can enroll in any course and other student functionality.

So in this way inheritance is a very powerful concept that allows you to reuse the code. It is the differece between a senior and a juior programmer.


What gets Inherited?
- Constructor of parent class
- Non Private Attributes of parent class
- Non Private Methods of parent class



```python
class user:
  def __init__(self):
    self.name = 'Muneeb'

  def login(self):
    print('login')


class student(user):   # Here i created the studet class and inherit it from user class.

  def enroll(self):
    print('Enrolled in the course')


u = user() # object of user class
s = student()

print(s.name)
s.login() # So you can see i can access the attributes and methods of parent class with the help of object of student class

# This is the implementation of Inheritance concept of the OOP.
```

## **Method Overriding:**

#### If two mehods are present in both parent and child class with same name and if we call it from the object of child class then that method will be called which is in the child class..

Link it with philosphy that if your parent want to make you doctor and you are more passionate and want a engineer then you will persue Engineering.

**Infact we have studied this above like when both parent and child class have constructor then the constructor of parent class was ignored and that concept was called as *Constructor Overriding* .**

```python
# Method Overriding
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")

s=SmartPhone(20000, "Apple", 13)

s.buy()
```

## **Super Keyword:**

A useful and powerful keyword and mostly asked in the interview.

**So If both parent and child have method with same name then when we create the object of the child and access that method then the child method will be called and we can acccess the method of parent by using the super keyword.**


You can see in the below example:

```python
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        # syntax to call parent ka buy method
        super().buy()

s=SmartPhone(20000, "Apple", 13)

s.buy()
```

```python
# Now this is a beautiful example so that you can understand where we use super and its benefit.
# Each mobile has price , brand but only smartphone has andriod and ram

# super -> constuctor
class Phone:
    def __init__(self, price, brand):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand

class SmartPhone(Phone):
    def __init__(self, price, brand, os, ram):
        print('Inside smartphone constructor')
        super().__init__(price, brand)
        self.os = os
        self.ram = ram
        print ("Inside smartphone constructor")

s=SmartPhone(20000, "Samsung", "Android", 2)

print(s.os)
print(s.brand)
```

## **Inheritance in summary**

- A class can inherit from another class.

- Inheritance improves code reuse

- Constructor, attributes, methods get inherited to the child class

- The parent has no access to the child class

- Private properties of parent are not accessible directly in child class

- Child class can override the attributes or methods. This is called method overriding

- super() is an inbuilt function which is used to invoke the parent class methods and constructor

### **Below are some examples to make the concept clear**

```python
class Parent:

    def __init__(self,num):
      self.__num=num

    def get_num(self):
      return self.__num

class Child(Parent):

    def __init__(self,num,val):
      super().__init__(num)
      self.__val=val

    def get_val(self):
      return self.__val

son=Child(100,200)
print(son.get_num())
print(son.get_val())
```

```python
class Parent:
    def __init__(self):
        self.num=100

class Child(Parent):

    def __init__(self):
        super().__init__()
        self.var=200

    def show(self):
        print(self.num) # here as we know self is also object to so, both son (object of child class) and self are same
        print(self.var) # Thats why it can access var of parent class

son=Child()
son.show()
```

```python
class Parent:
    def __init__(self):
        self.__num=100

    def show(self):
        print("Parent:",self.__num)

class Child(Parent):
    def __init__(self):
        super().__init__()
        self.__var=10

    def show(self):
        print("Child:",self.__var)

obj=Child()
obj.show()
```

### **Types of Inheritance**

- **Single Inheritance:** one child one parent
- **Multilevel Inheritance:** infinite series of parents and childs
- **Hierarchical Inheritance:** one parent and multiple childs
- **Multiple Inheritance(Diamond Problem):** multiple parents and one child
- **Hybrid Inheritance:** combo of different types of inheritance


Below are the examples of each type of inheritance:

```python
# single inheritance
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

SmartPhone(1000,"Apple","13px").buy()
```

```python
# multilevel
class Product:
    def review(self):
        print ("Product customer review")

class Phone(Product):
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
s.review()
```

```python
# Hierarchical
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

class FeaturePhone(Phone):
    pass

SmartPhone(1000,"Apple","13px").buy()
FeaturePhone(10,"Lava","1px").buy()
```

```python
# Multiple
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def review(self):
        print ("Customer review")

class SmartPhone(Phone, Product): # This child is inheriting both parent classes.
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
s.review()
```

```python
# the diamond problem
# https://stackoverflow.com/questions/56361048/what-is-the-diamond-problem-in-python-and-why-its-not-appear-in-python2
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def buy(self):
        print ("Product buy method")

# Method resolution order
class SmartPhone(Phone,Product): # here we write Phone class pehly then Product so Phone will have priority and that why phone class's buy method will be called.
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
```

##**Method Overriding:**

## **Method Overloading:**
When we have two with same name in a class but both of them behave differently according to the input because they have different number of parameters.
eg: we have two function for calculaing area. one is def area(self, radius) to calculate area of circle and the other is def area(self, lenght, width).

But Method Overloading doesnot work in python, in java it works. and if mistakenly we do this then python will take or look only 2nd method and ignore upper one. AND if we want to do method overloading python has already a powerful concept for this which is keyword arguments.

## **Operator Overloading:**
When we have same operator but its behaviour is different based on different inputs. like when we make the fraction class and write the logic of + operator that adds two fractions. so to implement operator overriding we use magic methods.

```python
# Method Overloading
# will cause error

class Area:
  def area(self, radius):
    return 3.14*radius*radius

  def area(self,lenght, width):
    return lenght*width

a = Area()
a.area(1)
```

```python
# Now how will tackle overloading in python


class Area:
  def area(self, a, b=0):
    if b ==0:
      return 3.14*a*a
    else:
      return a*b

a = Area()
print(a.area(1))
print(a.area(2,3))
```
